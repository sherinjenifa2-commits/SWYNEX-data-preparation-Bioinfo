The GSE2034 log2-transformed dataset contains 22,283 probes and 286 numeric sample columns. Data quality assessment identified no missing values, duplicate rows, duplicate probe IDs, or infinite values. All 286 sample columns were stored as float64, while the probe identifier column was stored as an object data type.

```python
# Select numerical sample columns
numeric_data = df.select_dtypes(include=np.number)

# Calculate statistical measures
print("Overall Mean:", numeric_data.mean().mean())

print("Overall Median:", numeric_data.median().median())

print("Overall Standard Deviation:",
      numeric_data.stack().std())
```

```python
# Calculate average expression for each probe
probe_mean = numeric_data.mean(axis=1)

# Top 10 probes
top_10 = probe_mean.sort_values(
    ascending=False
).head(10)

print("Top 10 Highly Expressed Probes:")
print(top_10)
```

```python
lowest_10 = probe_mean.sort_values(
    ascending=True
).head(10)

print("10 Lowest Expressed Probes:")
print(lowest_10)
```

```python
sample_mean = numeric_data.mean()

print("Sample-wise Mean Expression:")
print(sample_mean.describe())
```

```python
plt.figure(figsize=(10, 5))

plt.hist(
    numeric_data.values.flatten(),
    bins=50
)

plt.title("Distribution of Gene Expression Values")
plt.xlabel("Log2 Expression Value")
plt.ylabel("Frequency")

plt.show()
```

The histogram shows the distribution of log2 gene expression values across the dataset. Most expression values are concentrated around the middle range, while fewer values occur at the lower and higher extremes. This helps us understand the overall expression pattern in the dataset.

```python
# Calculate correlation between samples
sample_corr = numeric_data.corr()

# Create heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    sample_corr,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Sample-to-Sample Correlation Heatmap")
plt.xlabel("Samples")
plt.ylabel("Samples")
plt.show()
```

Sample-to-Sample Correlation Analysis
The correlation heatmap shows the relationship between the 286 samples based on their gene expression profiles. Most cells appear red, indicating positive correlations between many samples. The diagonal line represents a perfect correlation of 1.0 because each sample is fully correlated with itself. This analysis helps identify similarities and differences among samples and can support further sample grouping and quality assessment.

```python
plt.figure(figsize=(15, 6))

sns.boxplot(data=numeric_data, showfliers=False)

plt.title("Gene Expression Distribution Across Samples")
plt.xlabel("Samples")
plt.ylabel("Log2 Expression Values")

plt.xticks([])
plt.show()
```

Boxplot Analysis
The boxplot compares gene expression distributions across all 286 samples. The median expression values and interquartile ranges appear broadly similar across samples, although some variation is present. The whiskers show the spread of expression values. This visualization helps compare sample distributions and identify samples that may require further investigation.

```python
# Calculate mean expression for each sample
sample_means = numeric_data.mean(axis=0)

# Plot sample-wise mean expression
plt.figure(figsize=(14, 5))

plt.plot(sample_means.values)

plt.title("Mean Gene Expression Across Samples")
plt.xlabel("Sample Number")
plt.ylabel("Mean Log2 Expression")

plt.show()
```

Sample-wise Mean Expression Analysis
The line plot shows the average gene expression values across 286 samples. The mean expression values range approximately from 7.42 to 8.07. Most samples have mean expression values between 7.6 and 7.9, indicating relatively small variation in average expression across samples. This comparison helps identify samples with unusually high or low average expression for further investigation.

```python
# Create average expression table
probe_analysis = pd.DataFrame({
    "Probe_ID": df["ID_REF"],
    "Mean_Expression": numeric_data.mean(axis=1)
})

# Top 10 probes
print("Top 10 probes:")
print(
    probe_analysis.sort_values(
        by="Mean_Expression",
        ascending=False
    ).head(10)
)

# Lowest 10 probes
print("Lowest 10 probes:")
print(
    probe_analysis.sort_values(
        by="Mean_Expression",
        ascending=True
    ).head(10)
)
```

The top 10 probes have the highest average expression values across all samples, while the lowest 10 probes have the lowest average expression values. These results help identify probes with relatively high and low expression levels in the dataset.

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Transpose data: samples become rows
X = numeric_data.T

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(X_scaled)

# Plot PCA
plt.figure(figsize=(8, 6))

plt.scatter(
    pca_result[:, 0],
    pca_result[:, 1]
)

plt.title("PCA of Gene Expression Samples")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```

PCA Analysis
Principal Component Analysis (PCA) was used to visualize the 286 gene expression samples in two dimensions. Each point represents one sample. The samples are spread across the plot, and no clearly separated clusters are visible. This visualization helps explore overall similarities and differences between samples. Further analysis with biological sample labels would be needed to interpret any possible grouping.

```python
all_values = numeric_data.to_numpy().flatten()

print("Exact Overall Mean:", np.mean(all_values))
print("Exact Overall Median:", np.median(all_values))
print("Exact Overall Standard Deviation:", np.std(all_values, ddof=1))
```

Interpretation:
The overall mean expression value is approximately 7.77, and the median is approximately 7.91. The standard deviation of 2.50 indicates variation in gene expression values across the dataset.
